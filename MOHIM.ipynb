```python
class Separator:
    def __init__(self, device="cuda"):
        self.model = get_model("htdemucs_6s")
        self.model.to(device)
        self.model.eval()
        self.device = device
        self.sr = self.model.samplerate

    @torch.no_grad()
    def separate(self, audio_path):
        wav, sr = torchaudio.load(audio_path)
        if wav.shape[0] == 1:
            wav = wav.repeat(2, 1)
        if sr != self.sr:
            wav = torchaudio.transforms.Resample(sr, self.sr)(wav)

        wav = wav.to(self.device)
        sources = apply_model(self.model, wav[None], device=self.device)[0]
        # sources 순서는 self.model.sources 참고 (기본: drums, bass, other, vocals)
        stems = {name: sources[i].cpu() for i, name in enumerate(self.model.sources)}
        return stems, self.sr
```

```python
detector = SwiftF0(fmin=46.875, fmax=2093.75, confidence_threshold=0.9)

def extract_notes(wav_mono, sr):
    """
    wav_mono: 1D float32 numpy array
    반환: PitchResult, NoteSegment 리스트
    """
    result = detector.detect_from_array(wav_mono, sr)
    notes = segment_notes(
        result,
        split_semitone_threshold=0.8,   # 이 이하 변화는 같은 음으로 취급 (어택 노이즈 방어)
        min_note_duration=0.05,         # 50ms 미만 잡음성 세그먼트 제거
        unvoiced_grace_period=0.05,     # pluck 사이 짧은 무음은 note 끊김으로 안 봄
    )
    return result, notes
```

```python
def movement_score(notes, octave_fold=True):
    """
    notes: List[NoteSegment] (SwiftF0.segment_notes 결과)
    """
    if len(notes) < 2:
        return 0.0

    midi_seq = [n.pitch_midi for n in notes]
    intervals = np.array([midi_seq[i+1] - midi_seq[i] for i in range(len(midi_seq)-1)])

    if octave_fold:
        # 옥타브 트래킹 에러 방지: interval을 [-6,5] 범위로 접음
        intervals = ((intervals + 6) % 12) - 6

    abs_intervals = np.abs(intervals)
    moved_ratio = np.mean(abs_intervals > 0)                       # 실제 음이 바뀐 전환 비율
    avg_interval = np.clip(np.mean(abs_intervals) / 5.0, 0, 1)     # 평균 변화폭

    pitch_classes = [m % 12 for m in midi_seq]
    unique_pc_ratio = len(set(pitch_classes)) / len(pitch_classes)  # 같은 음 반복이면 낮음

    return float(0.4 * moved_ratio + 0.3 * avg_interval + 0.3 * unique_pc_ratio)

def repetition_score(notes, min_ngram=3, max_ngram=6, octave_fold=True):
    if len(notes) < min_ngram + 1:
        return 0.0

    midi_seq = [n.pitch_midi for n in notes]
    intervals = [midi_seq[i+1] - midi_seq[i] for i in range(len(midi_seq)-1)]
    if octave_fold:
        intervals = [((iv + 6) % 12) - 6 for iv in intervals]

    covered = np.zeros(len(intervals), dtype=bool)
    total_weight, matched_weight = 0.0, 0.0

    for n in range(min_ngram, min(max_ngram, len(intervals)) + 1):
        positions = {}
        for i in range(len(intervals) - n + 1):
            key = tuple(intervals[i:i+n])
            positions.setdefault(key, []).append(i)
        for key, pos_list in positions.items():
            total_weight += n
            if len(pos_list) > 1:
                matched_weight += n
                for p in pos_list:
                    covered[p:p+n] = True

    if total_weight == 0:
        return 0.0

    return float(0.5 * (matched_weight / total_weight) + 0.5 * covered.mean())

def score_motif_candidate(stem_wav, full_wav, sr, hop_length=512):
    stem = stem_wav.mean(0).numpy()
    full = full_wav.mean(0).numpy()
    full_peak = np.max(np.abs(full))
    stem = stem / full_peak
    full = full / full_peak

    stem_rms = librosa.feature.rms(y=stem, hop_length=hop_length)[0]
    full_rms = librosa.feature.rms(y=full, hop_length=hop_length)[0]
    active = full_rms > 1e-4
    contribution = stem_rms[active] / (full_rms[active] + 1e-8)
    presence = 0.4 * np.mean(contribution) + 0.6 * np.percentile(contribution, 95)

    if presence < 0.3:
        movement, repetition = 0.0, 0.0
    else:
        _, notes = extract_notes(stem.astype(np.float32), sr)
        movement = movement_score(notes)
        repetition = repetition_score(notes)

    # presence 가중치를 낮춰서 "믹스에서 크게 들림"이 아니라 "멜로디 움직임/반복"이 승부를 가르게
    total = 0.5 * presence + 0.25 * movement + 0.25 * repetition

    return {
        "presence": float(presence),
        "movement": float(movement),
        "repetition": float(repetition),
        "total": float(total),
    }

def select_motif_stem(stems, full_wav, sr, priority_order=("guitar", "piano", "bass", "other")):
    scores = {}
    for name in priority_order:
        if name not in stems:
            continue
        scores[name] = score_motif_candidate(stems[name], full_wav, sr)

    if not scores:
        raise ValueError("No valid stems to score")

    best = max(scores.items(), key=lambda kv: kv[1]["total"])
    return best[0], scores
```

```python
def get_downbeats(audio_path, beat_tracker):
    beats, downbeats = beat_tracker(audio_path)
    beats = np.asarray(beats, dtype=np.float64)
    downbeats = np.asarray(downbeats, dtype=np.float64)
    return beats, downbeats

def process_song(audio_path, separator):
    stems, sr = separator.separate(audio_path)
    beats, downbeats = beat_tracker(audio_path)
    beats = np.asarray(beats, dtype=np.float64).reshape(-1)
    downbeats = np.asarray(downbeats, dtype=np.float64).reshape(-1)
    return stems, sr, beats, downbeats

def resize_time(feature, target_frames):
    old_x = np.linspace(0, 1, feature.shape[1])
    new_x = np.linspace(0, 1, target_frames)
    return np.vstack([
        np.interp(new_x, old_x, row)
        for row in feature
    ])

# def extract_best_motif_segment_downbeat(stem_wav, sr, downbeats, n_bars=4, fix_frames=64,
#                                             search_sec=30, similarity_threshold=0.56):
#     mono = stem_wav.mean(0).numpy()
#     n_db = len(downbeats)
#     if n_db <= n_bars * 2:
#         return None, None

#     hop_length = 512
#     bar_len = np.median(np.diff(downbeats))
#     motif_len = bar_len * n_bars
#     onset_env = librosa.onset.onset_strength(y=mono, sr=sr, hop_length=hop_length)
#     chroma = librosa.feature.chroma_cens(y=mono, sr=sr, hop_length=hop_length)

#     segments = []
#     for s in downbeats:
#         e = s + motif_len
#         if e * sr > len(mono):
#             continue

#         start_frame = int(librosa.time_to_frames(s, sr=sr, hop_length=hop_length))
#         end_frame = int(librosa.time_to_frames(e, sr=sr, hop_length=hop_length))

#         segment_onset = resize_time(onset_env[None, start_frame:end_frame], fix_frames).flatten()
#         segment_chroma = resize_time(chroma[:, start_frame:end_frame], fix_frames).flatten()
#         segments.append((s, e, segment_onset, segment_chroma))

#     selected = None

#     for s, e, first_onset, first_chroma in segments:
#         if s >= search_sec:
#             break

#         pair_scores = []

#         for other_s, _, other_onset, other_chroma in segments:
#             if abs(other_s - s) < motif_len * 0.9:
#                 continue

#             onset_sim = cosine_similarity(
#                 first_onset[None],
#                 other_onset[None],
#             )[0, 0]

#             chroma_sim = cosine_similarity(
#                 first_chroma[None],
#                 other_chroma[None],
#             )[0, 0]

#             score = 0.3 * onset_sim + 0.7 * chroma_sim

#             pair_scores.append(
#                 (other_s, onset_sim, chroma_sim, score)
#             )

#         if not pair_scores:
#             continue

#         onset_mean = np.mean([
#             onset_sim
#             for _, onset_sim, _, _ in pair_scores
#         ])

#         chroma_mean = np.mean([
#             chroma_sim
#             for _, _, chroma_sim, _ in pair_scores
#         ])

#         score_mean = np.mean([
#             score
#             for _, _, _, score in pair_scores
#         ])

#         best_other_s, best_onset, best_chroma, best_score = max(
#             pair_scores,
#             key=lambda item: item[3],
#         )

#         print(
#             f"start={s:.3f} | "
#             f"onset_mean={onset_mean:.3f} | "
#             f"chroma_mean={chroma_mean:.3f} | "
#             f"score_mean={score_mean:.3f} | "
#             f"best_match={best_other_s:.3f} | "
#             f"best_onset={best_onset:.3f} | "
#             f"best_chroma={best_chroma:.3f} | "
#             f"best_score={best_score:.3f}"
#         )

#         if selected is None and score_mean >= similarity_threshold:
#             selected = (int(s * sr), int(e * sr))

#     if selected is not None:
#         return selected

#     return None, None

def extract_best_motif_segment_downbeat(stem_wav, sr, downbeats, n_bars=4, fix_frames=64,
                                            search_sec=30, similarity_threshold=0.56,
                                            silence_db=-40):
    mono = stem_wav.mean(0).numpy()
    n_db = len(downbeats)
    if n_db <= n_bars * 2:
        return None, None

    hop_length = 512
    bar_len = np.median(np.diff(downbeats))
    motif_len = bar_len * n_bars
    onset_env = librosa.onset.onset_strength(y=mono, sr=sr, hop_length=hop_length)
    chroma = librosa.feature.chroma_cens(y=mono, sr=sr, hop_length=hop_length)
    rms = librosa.feature.rms(y=mono, hop_length=hop_length)[0]
    rms_db = librosa.amplitude_to_db(rms + 1e-8, ref=np.max)

    segments = []
    for s in downbeats:
        e = s + motif_len
        if e * sr > len(mono):
            continue

        start_frame = int(librosa.time_to_frames(s, sr=sr, hop_length=hop_length))
        end_frame = int(librosa.time_to_frames(e, sr=sr, hop_length=hop_length))

        segment_onset = resize_time(onset_env[None, start_frame:end_frame], fix_frames).flatten()
        segment_chroma = resize_time(chroma[:, start_frame:end_frame], fix_frames).flatten()
        segment_rms = rms_db[start_frame:end_frame]
        active = segment_rms > silence_db

        if np.any(active):
            first_active = np.flatnonzero(active)[0]
            leading_silence_ratio = first_active / len(active)
        else:
            leading_silence_ratio = 1.0

        segments.append((s, e, segment_onset, segment_chroma, leading_silence_ratio))

    for s, e, first_onset, first_chroma, leading_silence_ratio in segments:
        if s >= search_sec:
            break

        match_scores = []
        for other_s, _, other_onset, other_chroma, _ in segments:
            if abs(other_s - s) < motif_len * 0.9:
                continue

            onset_sim = cosine_similarity(first_onset[None], other_onset[None])[0, 0]
            chroma_sim = cosine_similarity(first_chroma[None], other_chroma[None])[0, 0]
            score = 0.3 * onset_sim + 0.7 * chroma_sim
            match_scores.append(score)

        mean_score = np.mean(match_scores) if match_scores else -1.0
        adjusted_score = mean_score * (1.0 - leading_silence_ratio)
        if adjusted_score >= similarity_threshold:
            return int(s * sr), int(e * sr)

    return None, None
```

```python
def pack_sequences(hidden1, hidden2, mask1, mask2):
    hidden_cat = torch.cat([hidden1, hidden2], dim=1)
    mask_cat = torch.cat([mask1, mask2], dim=1)
    B, L, D = hidden_cat.shape
    sort_idx = mask_cat.argsort(dim=1, descending=True, stable=True)
    hidden_left = torch.gather(hidden_cat, 1, sort_idx.unsqueeze(-1).expand(B, L, D))
    lengths = mask_cat.sum(dim=1)
    new_mask = (
        torch.arange(L, dtype=torch.long, device=hidden_cat.device).unsqueeze(0)
        < lengths.unsqueeze(1)
    )
    return hidden_left, new_mask

class MotifConditionedAceStep15(nn.Module):
    """
    AceStepConditionGenerationModel을 감싸서, 모티프의 프레임 단위 acoustic
    latent를 pooling 없이 encoder_hidden_states에 4번째 조건으로 추가한다.
    """

    def __init__(self, gen_model, lora_rank=16, lora_alpha=32):
        super().__init__()
        self.gen_model = gen_model  # AceStepConditionGenerationModel 인스턴스

        for p in self.gen_model.parameters():
            p.requires_grad_(False)

        # --- LoRA 부착: q_proj/k_proj/v_proj/o_proj (실제 확인된 이름) ---
        target_modules = self._find_target_modules(self.gen_model.decoder)
        lora_config = LoraConfig(
            r=lora_rank,
            lora_alpha=lora_alpha,
            target_modules=target_modules,
            lora_dropout=0.0,
        )
        self.gen_model.decoder = get_peft_model(self.gen_model.decoder, lora_config)

        # --- 모티프 프레임을 encoder_hidden_size로 projection하는 새 레이어 ---
        # AceStepConditionEncoder는 encoder_config(별도 hidden_size)를 쓰므로
        # 그 값을 그대로 재사용 (XL 모델은 decoder inner_dim과 다를 수 있음)
        encoder_hidden_size = self.gen_model.encoder.text_projector.out_features
        audio_acoustic_dim = self.gen_model.tokenizer.audio_acoustic_proj.in_features

        self.motif_projector = nn.Linear(audio_acoustic_dim, encoder_hidden_size, bias=False)
        # 모티프 토큰임을 구분하는 학습 가능 임베딩 (lyric/timbre/text와 구분용)
        self.motif_type_embedding = nn.Parameter(torch.randn(1, 1, encoder_hidden_size) * 0.02)

    @staticmethod
    def _find_target_modules(decoder):
        """decoder 안에서 q_proj/k_proj/v_proj/o_proj를 갖는 모듈 이름을 스캔."""
        names = []
        for name, module in decoder.named_modules():
            if any(name.endswith(p) for p in ["q_proj", "k_proj", "v_proj", "o_proj"]):
                names.append(name)
        return names

    def encode_motif(self, motif_acoustic_latent, motif_attention_mask):
        """
        motif_acoustic_latent: [B, T_motif, audio_acoustic_hidden_dim] (pooling 없음)
        motif_attention_mask: [B, T_motif]
        """
        motif_tokens = self.motif_projector(motif_acoustic_latent)
        motif_tokens = motif_tokens + self.motif_type_embedding
        return motif_tokens, motif_attention_mask

    def build_encoder_hidden_states(
        self,
        text_hidden_states, text_attention_mask,
        lyric_hidden_states, lyric_attention_mask,
        refer_audio_acoustic_hidden_states_packed, refer_audio_order_mask,
        motif_acoustic_latent, motif_attention_mask,
    ):
        enc = self.gen_model.encoder

        text_proj = enc.text_projector(text_hidden_states)

        lyric_out = enc.lyric_encoder(
            inputs_embeds=lyric_hidden_states, attention_mask=lyric_attention_mask,
        )
        lyric_hidden = lyric_out.last_hidden_state

        timbre_unpack, timbre_mask = enc.timbre_encoder(
            refer_audio_acoustic_hidden_states_packed, refer_audio_order_mask,
        )

        motif_tokens, motif_mask = self.encode_motif(motif_acoustic_latent, motif_attention_mask)

        # lyric + timbre + text 순서(원본과 동일)로 먼저 패킹한 뒤, 모티프를 4번째로 추가
        encoder_hidden_states, mask = pack_sequences(lyric_hidden, timbre_unpack, lyric_attention_mask, timbre_mask)
        encoder_hidden_states, mask = pack_sequences(encoder_hidden_states, text_proj, mask, text_attention_mask)
        encoder_hidden_states, mask = pack_sequences(encoder_hidden_states, motif_tokens, mask, motif_mask)

        return encoder_hidden_states, mask

    def forward(
        self,
        hidden_states, attention_mask,
        text_hidden_states, text_attention_mask,
        lyric_hidden_states, lyric_attention_mask,
        refer_audio_acoustic_hidden_states_packed, refer_audio_order_mask,
        motif_acoustic_latent, motif_attention_mask,
        src_latents, chunk_masks, is_covers, silence_latent,
        cfg_ratio=0.15,
    ):
        encoder_hidden_states, encoder_attention_mask = self.build_encoder_hidden_states(
            text_hidden_states, text_attention_mask,
            lyric_hidden_states, lyric_attention_mask,
            refer_audio_acoustic_hidden_states_packed, refer_audio_order_mask,
            motif_acoustic_latent, motif_attention_mask,
        )

        context_latents = torch.cat([src_latents, chunk_masks.to(src_latents.dtype)], dim=-1)

        bsz, device, dtype = hidden_states.shape[0], hidden_states.device, hidden_states.dtype

        # CFG dropout: 모티프를 포함한 조건 전체를 통째로 null로 대체
        cfg_mask = torch.where(
            torch.rand(size=(bsz,), device=device, dtype=dtype) < cfg_ratio,
            torch.zeros(size=(bsz,), device=device, dtype=dtype),
            torch.ones(size=(bsz,), device=device, dtype=dtype),
        ).view(-1, 1, 1)
        null_emb = self.gen_model.null_condition_emb.expand_as(encoder_hidden_states)
        encoder_hidden_states = torch.where(cfg_mask > 0, encoder_hidden_states, null_emb)

        # 1.5의 실제 rectified flow 공식 그대로
        x1 = torch.randn_like(hidden_states)
        x0 = hidden_states
        t = torch.sigmoid(torch.randn(bsz, device=device, dtype=dtype))
        t_ = t.unsqueeze(-1).unsqueeze(-1)
        xt = t_ * x1 + (1.0 - t_) * x0

        decoder_outputs = self.gen_model.decoder(
            hidden_states=xt,
            timestep=t,
            timestep_r=t,
            attention_mask=attention_mask,
            encoder_hidden_states=encoder_hidden_states,
            encoder_attention_mask=encoder_attention_mask,
            context_latents=context_latents,
        )

        flow = x1 - x0
        loss = F.mse_loss(decoder_outputs[0], flow)
        return loss

    def trainable_parameters(self):
        return [p for p in self.parameters() if p.requires_grad]

def build_index(out_dir):
    index = []
    for song_dir in sorted(os.listdir(out_dir)):
        song_path = os.path.join(out_dir, song_dir)
        meta_path = os.path.join(song_path, "metadata.json")
        if not os.path.isdir(song_path) or not os.path.exists(meta_path):
            continue

        with open(meta_path) as f:
            meta = json.load(f)

        motif_stem = meta.get("motif_stem")
        if motif_stem is None:
            continue

        motif_stem_path = os.path.join(song_path, f"{motif_stem}.wav")
        vocal_path = os.path.join(song_path, "vocals.wav")
        if not (os.path.exists(motif_stem_path) and os.path.exists(vocal_path)):
            continue

        index.append({
            "song_id": song_dir,
            "motif_stem_path": motif_stem_path,
            "motif_start_sec": meta["motif_start_sec"],
            "motif_end_sec": meta["motif_end_sec"],
            "vocal_path": vocal_path,
        })
    return index

class MotifVocalDataset15(Dataset):
    def __init__(self, index, target_sr=44100, max_target_sec=30.0):
        self.index = index
        self.target_sr = target_sr
        self.max_target_sec = max_target_sec

    def __len__(self):
        return len(self.index)

    def _load(self, path, start_sec=None, end_sec=None):
        info = torchaudio.info(path)
        sr = info.sample_rate
        if start_sec is not None:
            frame_offset = int(start_sec * sr)
            num_frames = int((end_sec - start_sec) * sr)
            wav, _ = torchaudio.load(path, frame_offset=frame_offset, num_frames=num_frames)
        else:
            wav, _ = torchaudio.load(path)
        if wav.shape[0] == 1:
            wav = wav.repeat(2, 1)
        if sr != self.target_sr:
            wav = torchaudio.transforms.Resample(sr, self.target_sr)(wav)
        return wav

    def __getitem__(self, idx):
        item = self.index[idx]

        motif_audio = self._load(
            item["motif_stem_path"], item["motif_start_sec"], item["motif_end_sec"]
        )
        target_audio = self._load(item["vocal_path"])

        max_len = int(self.max_target_sec * self.target_sr)
        if target_audio.shape[-1] > max_len:
            start = random.randint(0, target_audio.shape[-1] - max_len)
            target_audio = target_audio[:, start:start + max_len]

        if random.random() < 0.15:
            motif_audio = torch.zeros_like(motif_audio)

        return {
            "motif_audio": motif_audio,
            "target_audio": target_audio,
            "caption": "lead vocal melody",
            "lyrics": "",
        }

def collate_fn(batch):
    return batch
```

```python
def train_step(model, dit_handler, batch, optimizer, device="cuda"):
    """
    실제 handler mixin 함수로 검증된 전처리:
    - target 인코딩: dit_handler._encode_audio_to_latents(wav)  [C, T] -> [T_latent, C_latent]
    - 모티프 인코딩: dit_handler.tiled_encode(motif_audio) 재사용
      (infer_refer_latent와 동일한 인코더지만, timbre_encoder pooling은
      거치지 않고 raw acoustic latent를 그대로 model.encode_motif로 보냄)
    - text/lyric: dit_handler.infer_text_embeddings / infer_lyric_embeddings
    - plain(비-cover) 샘플의 src_latents/chunk_masks: silence_latent로 구성
      (conditioning_target.py의 실제 로직과 동일한 패턴)
    """
    model.train()
    optimizer.zero_grad()
    dit_handler._ensure_silence_latent_on_device()

    total_loss = 0.0
    for sample in batch:
        target_wav = sample["target_audio"].to(device)   # [2, T]
        motif_wav = sample["motif_audio"].to(device)      # [2, T_motif]

        with torch.inference_mode():
            # --- target latent (실제 handler 함수 재사용) ---
            target_latent = dit_handler._encode_audio_to_latents(target_wav)  # [T_lat, C]
            target_latent = target_latent.unsqueeze(0)  # [1, T_lat, C]

            # --- 모티프 latent: 같은 VAE, pooling 없이 raw latent 그대로 ---
            motif_latent = dit_handler.tiled_encode(motif_wav, offload_latent_to_cpu=True)
            motif_latent = motif_latent.to(device).to(target_latent.dtype)
            if motif_latent.dim() == 2:
                motif_latent = motif_latent.unsqueeze(0)
            motif_latent = motif_latent.transpose(1, 2)  # infer_refer_latent와 동일한 축 정렬
            motif_attention_mask = torch.ones(
                motif_latent.shape[0], motif_latent.shape[1], device=device
            )

            # --- text/lyric ---
            # TODO: 실제 토큰화는 handler의 prompt_utils.py/PromptMixin 쪽 토크나이저를
            # 거쳐야 함. 여기서는 이미 토큰화된 tensor가 있다고 가정.
            text_token_ids = sample["text_token_ids"].to(device)
            lyric_token_ids = sample["lyric_token_ids"].to(device)
            text_hidden_states = dit_handler.infer_text_embeddings(text_token_ids)
            lyric_hidden_states = dit_handler.infer_lyric_embeddings(lyric_token_ids)
            text_attention_mask = torch.ones(
                text_hidden_states.shape[0], text_hidden_states.shape[1], device=device
            )
            lyric_attention_mask = torch.ones(
                lyric_hidden_states.shape[0], lyric_hidden_states.shape[1], device=device
            )

            # --- refer_audio: 이번 학습에선 안 씀 (silence로 채움) ---
            silence_ref = dit_handler._get_silence_latent_slice(750).unsqueeze(0)
            refer_audio_order_mask = torch.tensor([0], device=device)

            # --- plain(non-cover) src_latents/chunk_masks: silence로 구성 ---
            latent_len = target_latent.shape[1]
            src_latents = dit_handler._get_silence_latent_slice(latent_len).unsqueeze(0)
            chunk_masks = torch.zeros(1, latent_len, device=device)
            is_covers = torch.zeros(1, device=device)
            attention_mask = torch.ones(1, latent_len, device=device)

        loss = model(
            hidden_states=target_latent,
            attention_mask=attention_mask,
            text_hidden_states=text_hidden_states,
            text_attention_mask=text_attention_mask,
            lyric_hidden_states=lyric_hidden_states,
            lyric_attention_mask=lyric_attention_mask,
            refer_audio_acoustic_hidden_states_packed=silence_ref,
            refer_audio_order_mask=refer_audio_order_mask,
            motif_acoustic_latent=motif_latent,
            motif_attention_mask=motif_attention_mask,
            src_latents=src_latents,
            chunk_masks=chunk_masks,
            is_covers=is_covers,
            silence_latent=dit_handler.silence_latent,
        )

        (loss / len(batch)).backward()
        total_loss += loss.item()

    torch.nn.utils.clip_grad_norm_(model.trainable_parameters(), max_norm=1.0)
    optimizer.step()
    return total_loss / len(batch)
```

```python
def tile_latents_to_length(latents, target_frame_length):
    """모티프 latent를 시간축으로 반복해서 목표 프레임 길이까지 채운다."""
    T = latents.shape[-1]
    n_repeat = math.ceil(target_frame_length / T)
    tiled = latents.repeat(1, 1, 1, n_repeat)
    return tiled[:, :, :, :target_frame_length]


def generate_section(
    pipeline,
    motif_path,
    section_prompt,
    section_lyrics,
    section_duration,
    noise_ratio,      # 0 = 모티프 그대로, 1 = 거의 완전히 새로 생성
    save_path,
    seed=seed,
    infer_step=60,
    batch_size=1,
):
    # 1) 모티프 인코딩
    motif_latents = pipeline.infer_latents(motif_path)  # [1, 8, 16, T_motif]

    # 2) 목표 섹션 길이(frame)까지 타일링
    target_frame_length = int(section_duration * 44100 / 512 / 8)
    ref_latents = tile_latents_to_length(motif_latents, target_frame_length)

    # 3) 텍스트/가사 임베딩 준비 (pipeline.__call__ 내부 로직과 동일하게)
    random_generators, actual_seeds = pipeline.set_seeds(batch_size, [seed])

    texts = [section_prompt]
    encoder_text_hidden_states, text_attention_mask = pipeline.get_text_embeddings(texts)
    encoder_text_hidden_states = encoder_text_hidden_states.repeat(batch_size, 1, 1)
    text_attention_mask = text_attention_mask.repeat(batch_size, 1)

    encoder_text_hidden_states_null = pipeline.get_text_embeddings_null(texts)
    encoder_text_hidden_states_null = encoder_text_hidden_states_null.repeat(batch_size, 1, 1)

    speaker_embeds = torch.zeros(batch_size, 512).to(pipeline.device).to(pipeline.dtype)

    lyric_token_idx = torch.tensor([0]).repeat(batch_size, 1).to(pipeline.device).long()
    lyric_mask = torch.tensor([0]).repeat(batch_size, 1).to(pipeline.device).long()
    if section_lyrics and len(section_lyrics) > 0:
        lyric_token_idx = pipeline.tokenize_lyrics(section_lyrics, debug=False)
        lyric_mask = [1] * len(lyric_token_idx)
        lyric_token_idx = torch.tensor(lyric_token_idx).unsqueeze(0).to(pipeline.device).repeat(batch_size, 1)
        lyric_mask = torch.tensor(lyric_mask).unsqueeze(0).to(pipeline.device).repeat(batch_size, 1)

    # 4) SDEdit(add_latents_noise)를 내부적으로 태우는 audio2audio 경로 직접 호출
    target_latents = pipeline.text2music_diffusion_process(
        duration=section_duration,               # ref_latents가 있으므로 실제로는 무시되지만 명시
        encoder_text_hidden_states=encoder_text_hidden_states,
        text_attention_mask=text_attention_mask,
        speaker_embds=speaker_embeds,
        lyric_token_ids=lyric_token_idx,
        lyric_mask=lyric_mask,
        random_generators=random_generators,
        infer_steps=infer_step,
        guidance_scale=15.0,
        omega_scale=10.0,
        scheduler_type="euler",
        cfg_type="apg",
        guidance_interval=0.5,
        guidance_interval_decay=0.0,
        min_guidance_scale=3.0,
        oss_steps=[],
        encoder_text_hidden_states_null=encoder_text_hidden_states_null,
        use_erg_lyric=True,
        use_erg_diffusion=True,
        guidance_scale_text=0.0,
        guidance_scale_lyric=0.0,
        audio2audio_enable=True,
        ref_audio_strength=1.0 - noise_ratio,   # 모티프 보존 비율
        ref_latents=ref_latents,
    )

    output_paths = pipeline.latents2audio(
        latents=target_latents,
        target_wav_duration_second=section_duration,
        save_path=save_path,
        format="wav",
    )
    return output_paths[0], actual_seeds[0]
```

```python
import math, torch

section_duration = 20.0  # 목표 섹션 길이 (모티프 9.6초보다 길게)
section_prompt = "acoustic pop, warm guitar, building energy, mid tempo, 100 bpm, gentle female vocal"
section_lyrics = "la la la la, la la la la\nla la la la, la la la la"

noise_ratios = [0.3, 0.6, 0.9]
section_results = {}

for ratio in noise_ratios:
    out_path = f"outputs/section_noise_{ratio}.wav"
    path, seed_used = generate_section(
        pipeline=pipeline,
        motif_path=motif_path,
        section_prompt=section_prompt,
        section_lyrics=section_lyrics,
        section_duration=section_duration,
        noise_ratio=ratio,
        save_path=out_path,
        seed= seed
    )
    section_results[ratio] = path
    print(f"noise_ratio={ratio} -> {path} (seed={seed_used})")
```
